## Step 2 — Worked Example *(10 minutes, instructor-led)*

A stock has a beta of 1.2 on the market factor and 0.5 on an interest-rate factor. The market risk premium is 6% and the interest-rate factor's risk premium is 2%. The risk-free rate is 4%. Find the stock's required return under this two-factor model.

```python
r_f = 0.04
beta_m1, rp_m = 1.2, 0.06
beta_r1, rp_r = 0.5, 0.02
```

```python
required1 = r_f + beta_m1 * rp_m + beta_r1 * rp_r
print(f'Required return = {required1:.2%}')
```

*Required return = 4% + 1.2×6% + 0.5×2% = 4%+7.2%+1% = 12.2%.*

## Step 4 — Python Pass *(7 minutes)*

- The multi-factor required-return formula, built so a third or fourth factor could be added just by adding a column
- Both stocks' required returns computed side by side
- A quick comparison row showing what CAPM alone (market factor only) would have estimated for each stock

```python
def multifactor_required_return(r_f, *factors):
    return r_f + sum(beta * rp for beta, rp in factors)

print(multifactor_required_return(r_f, (beta_m1, rp_m), (beta_r1, rp_r)))
print(multifactor_required_return(r_f, (beta_m2, rp_m), (beta_r2, rp_r)))
```
