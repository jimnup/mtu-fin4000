## Step 2 — Worked Example *(10 minutes, instructor-led)*

Five years of returns for Stock A and T-bills.

```python
returns_A = np.array([0.08, -0.03, 0.22, 0.15, -0.10])
returns_T = np.array([0.02,  0.02, 0.03, 0.03,  0.04])
```

```python
avg_A = returns_A.mean()
std_A = returns_A.std(ddof=1)
print(f'Avg={avg_A:.2%}  Std={std_A:.2%}')
```

## Step 4 — Python Pass *(7 minutes)*

- AVERAGE() and STDEV() for Stock A and T-bills
- A risk-premium formula that references those cells
- A Sharpe-ratio formula built from the cells above — not retyped numbers

```python
def summarize_returns(returns, riskfree):
    avg = returns.mean()
    std = returns.std(ddof=1)
    rp = avg - riskfree.mean()
    return {'avg': avg, 'std': std, 'risk_premium': rp, 'sharpe': rp / std}

summarize_returns(returns_A, returns_T)
```
