## Step 2 — Worked Example *(10 minutes, instructor-led)*

The risk-free rate is 4%, the market risk premium is 6%. A stock has beta 1.4. Find its CAPM-required return, and compare it to the stock's own forecast return of 12%.

```python
r_f, mrp = 0.04, 0.06
beta1, forecast1 = 1.4, 0.12
```

```python
required1 = r_f + beta1 * mrp
alpha1 = forecast1 - required1
print(f'Required={required1:.2%}  Alpha={alpha1:.2%}')
```

*Required return = 4%+1.4×6%=12.4%. Alpha = 12%−12.4% = −0.4%. This stock is forecast to earn slightly LESS than the CAPM says it should for its risk — it appears slightly overpriced.*

```python
beta2, forecast2 = 0.8, 0.10
required2 = r_f + beta2 * mrp
alpha2 = forecast2 - required2
print(f'Required={required2:.2%}  Alpha={alpha2:.2%}  ->', 'underpriced' if alpha2 > 0 else 'overpriced')
```

## Step 4 — Python Pass *(7 minutes)*

- The CAPM required-return formula, built from rƒ/beta/market-risk-premium cells
- Alpha computed automatically as forecast minus required return, for both stocks
- A small SML chart plotting both stocks' beta against required return

```python
import matplotlib.pyplot as plt
beta_range = np.linspace(0, 2, 50)
sml = r_f + beta_range * mrp

plt.figure(figsize=(6, 4))
plt.plot(beta_range, sml, label='Security Market Line')
plt.scatter([beta1, beta2], [forecast1, forecast2], color='red', zorder=5, label='Forecast returns')
plt.annotate('Stock 1', (beta1, forecast1)); plt.annotate('Stock 2', (beta2, forecast2))
plt.xlabel('Beta'); plt.ylabel('Expected return')
plt.title('Security Market Line'); plt.legend(); plt.grid(alpha=0.3); plt.show()
```
