## Step 2 — Worked Example *(10 minutes, instructor-led)*

Spot rates (zero-coupon yields) for Years 1-3.

```python
y1, y2, y3 = 0.04, 0.05, 0.055
```

```python
f2 = (1 + y2)**2 / (1 + y1) - 1
print(f'f2 = {f2:.2%}')
```

*f₂ = (1.05)² / 1.04 − 1 = 6.01%. Since the curve slopes upward, the implied forward rate is higher than either spot rate — the market is pricing in higher short-term rates ahead.*

## Step 3 — Pair Work *(12 minutes, with a partner)*

**1.** Compute f₃, the 1-year forward rate for Year 3, using y₂ and y₃.

```python
f3 = (1 + y3)**3 / (1 + y2)**2 - 1
print(f'f3 = {f3:.2%}')
```

**2.** Price a 3-year, 6% annual-coupon, $1,000 face-value bond by discounting each cash flow at its own maturity-matched spot rate.

```python
cash_flows = np.array([60, 60, 1060])
spot_rates = np.array([y1, y2, y3])
periods = np.array([1, 2, 3])
price_spot = (cash_flows / (1 + spot_rates) ** periods).sum()
print(f'Price (spot-rate discounting) = ${price_spot:.2f}')
```

**3.** Now price the same bond using a single flat rate equal to the 3-year spot rate (5.5%) for all three cash flows. How does this price compare to the one from spot-rate discounting?

```python
price_flat = (cash_flows / (1 + y3) ** periods).sum()
print(f'Price (flat {y3:.1%} rate) = ${price_flat:.2f}  (differs by ${price_spot-price_flat:.2f})')
```

## Step 4 — Python Pass *(7 minutes)*

- Forward-rate formulas (f₂ and f₃) built directly from the spot-rate cells
- Bond price using spot-rate discounting — each cash flow divided by its own (1+spot)^t
- The same bond priced with a single flat rate, for side-by-side comparison

```python
def price_with_spot_rates(cash_flows, rates, periods):
    return (cash_flows / (1 + rates) ** periods).sum()

def price_with_flat_rate(cash_flows, flat_rate, periods):
    return (cash_flows / (1 + flat_rate) ** periods).sum()

print(price_with_spot_rates(cash_flows, spot_rates, periods))
print(price_with_flat_rate(cash_flows, y3, periods))
```
