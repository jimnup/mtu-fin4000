## Step 2 — Worked Example *(10 minutes, instructor-led)*

Three stocks make up a small index. Stock Y then does a 4-for-1 split.

```python
# Given: a 3-stock index, before and after a 4-for-1 split in Stock Y
prices_day0  = np.array([20, 80, 40])   # X, Y, Z
shares_day0  = np.array([10,  2,  5])   # millions
prices_after = np.array([20, 20, 40])   # Y split 4-for-1: $80 -> $20
shares_after = np.array([10,  8,  5])   # Y's shares: 2M -> 8M
```

**Price-weighted index** — average of the raw prices (divisor = 3, unadjusted):

```python
pw_day0  = prices_day0.mean()
pw_after = prices_after.mean()
print(f'PW day0={pw_day0:.2f}   PW after (distorted)={pw_after:.2f}')
```

**Value-weighted index** — based on total market cap, which a split doesn't change:

```python
mktcap_day0  = (prices_day0 * shares_day0).sum()
mktcap_after = (prices_after * shares_after).sum()
vw_after = mktcap_after / mktcap_day0 * 100
print(f'Market cap day0={mktcap_day0}  after={mktcap_after}  VW index after={vw_after:.2f}')
```

**1.** Compute the value-weighted (market-cap-weighted) index level at Day 0 — set it to 100.

**2.** Compute the value-weighted index level right after the split. What changed, and why?

```python
face, price, days = 10000, 9800, 91
bank_discount_yield = (face - price) / face * (360 / days)
bond_equiv_yield    = (face - price) / price * (365 / days)
print(f'Bank discount yield={bank_discount_yield:.4%}  BEY={bond_equiv_yield:.4%}')
```

## Step 4 — Python Pass *(7 minutes)*

Wrap the two index calculations into reusable functions:

- The price-weighted index and its divisor adjustment after the split
- The value-weighted index (market cap Day 0 vs. after split)
- Both T-bill yield formulas from one Face/Price/Days input block

```python
def price_weighted(prices, divisor=None):
    return prices.sum() / (divisor if divisor else len(prices))

def value_weighted(prices, shares, base_prices, base_shares):
    return (prices * shares).sum() / (base_prices * base_shares).sum() * 100

print(price_weighted(prices_after, divisor=new_divisor))
print(value_weighted(prices_after, shares_after, prices_day0, shares_day0))
```
