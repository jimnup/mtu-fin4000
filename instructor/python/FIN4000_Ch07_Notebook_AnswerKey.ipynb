## Step 2 — Worked Example *(10 minutes, instructor-led)*

Combine two stocks into a portfolio. Compute the portfolio's expected return and standard deviation for a 50/50 mix.

```python
E_rA, sigma_A = 0.10, 0.20
E_rB, sigma_B = 0.15, 0.30
rho = 0.20
wA, wB = 0.5, 0.5
```

```python
E_rp = wA * E_rA + wB * E_rB
var_p = wA**2*sigma_A**2 + wB**2*sigma_B**2 + 2*wA*wB*sigma_A*sigma_B*rho
sigma_p = np.sqrt(var_p)
print(f'E(rp)={E_rp:.2%}  sigma_p={sigma_p:.2%}  (below either stock alone!)')
```

*E(rₚ) = 0.5(10%)+0.5(15%) = 12.5%. σₚ² = 0.25(0.04)+0.25(0.09)+2(0.5)(0.5)(0.20)(0.30)(0.2) = 0.0385 → σₚ = 19.62%. Notice this is BELOW either stock's own standalone risk (20% and 30%) — that's diversification at work.*

## Step 3 — Pair Work *(12 minutes, with a partner)*

**1.** Recompute σₚ using correlation ρ = −0.2 instead of +0.2, same weights. How much did the risk drop?

```python
rho_neg = -0.2
var_p_neg = wA**2*sigma_A**2 + wB**2*sigma_B**2 + 2*wA*wB*sigma_A*sigma_B*rho_neg
sigma_p_neg = np.sqrt(var_p_neg)
print(f'sigma_p (rho=-0.2) = {sigma_p_neg:.2%}')
```

**2.** Recompute σₚ using correlation ρ = 1.0 (perfectly correlated). What happened to the diversification benefit?

```python
rho_one = 1.0
var_p_one = wA**2*sigma_A**2 + wB**2*sigma_B**2 + 2*wA*wB*sigma_A*sigma_B*rho_one
sigma_p_one = np.sqrt(var_p_one)
print(f'sigma_p (rho=1.0) = {sigma_p_one:.2%}  (= simple weighted average, no diversification)')
```

**3.** Find the minimum-variance weights: wᴀ = (σʙ²−σᴀσʙρ) ÷ (σᴀ²+σʙ²−2σᴀσʙρ), using the original ρ = 0.2. What's the resulting portfolio's risk?

```python
wA_min = (sigma_B**2 - sigma_A*sigma_B*rho) / (sigma_A**2 + sigma_B**2 - 2*sigma_A*sigma_B*rho)
wB_min = 1 - wA_min
sigma_min = np.sqrt(wA_min**2*sigma_A**2 + wB_min**2*sigma_B**2 + 2*wA_min*wB_min*sigma_A*sigma_B*rho)
print(f'wA_min={wA_min:.2%}  sigma_min={sigma_min:.2%}')
```

## Step 4 — Python Pass *(7 minutes)*

- The portfolio return and variance formulas, built from weight/return/σ/correlation cells so any correlation can be tested
- A correlation sensitivity check: −1, −0.2, 0, +0.2, +1, watching portfolio risk change with everything else held fixed
- The minimum-variance weight formula, and the resulting portfolio's risk

```python
rho_values = np.array([-1, -0.2, 0, 0.2, 1])
var_p_values = wA**2*sigma_A**2 + wB**2*sigma_B**2 + 2*wA*wB*sigma_A*sigma_B*rho_values
sigma_p_values = np.sqrt(var_p_values)
for r, s in zip(rho_values, sigma_p_values):
    print(f'rho={r:>5.1f}:  sigma_p = {s:.2%}')
```
