## Step 2 — Worked Example *(10 minutes, instructor-led)*

A portfolio earned 14% with a standard deviation of 20% and a beta of 1.1. The risk-free rate is 4% and the market returned 11%. Compute the Treynor measure and Jensen's alpha.

```julia
r_f, r_m = 0.04, 0.11
rp_A, sigma_A, beta_A = 0.14, 0.20, 1.1
```

```julia
treynor_A = (rp_A - r_f) / beta_A
required_A = r_f + beta_A * (r_m - r_f)
jensen_A = rp_A - required_A
println(string("Treynor=", @sprintf("%.2f%%", (treynor_A) * 100), "  Required=", @sprintf("%.2f%%", (required_A) * 100), "  Jensen alpha=", @sprintf("%.2f%%", (jensen_A) * 100)))
```

*Treynor = (14%−4%)÷1.1 = 9.09%. Required return = 4%+1.1×(11%−4%) = 11.7%. Jensen's alpha = 14%−11.7% = 2.3%. This manager beat their risk-adjusted benchmark by 2.3 percentage points.*

## Step 3 — Pair Work *(12 minutes, with a partner)*

**1.** Compute this portfolio's Sharpe ratio too: (rₚ−rƒ)÷σₚ — the same formula from Chapter 5, applied to a manager's realized return.

```julia
sharpe_A = (rp_A - r_f) / sigma_A
println(string("Sharpe (A) = ", @sprintf("%.2f", sharpe_A)))
```

**2.** A second manager earned 12% with beta 0.8 and standard deviation 25%. Compute their Treynor measure, Jensen's alpha, and Sharpe ratio.

```julia
rp_B, sigma_B, beta_B = 0.12, 0.25, 0.8
treynor_B = (rp_B - r_f) / beta_B
required_B = r_f + beta_B * (r_m - r_f)
jensen_B = rp_B - required_B
sharpe_B = (rp_B - r_f) / sigma_B
println(string("Treynor=", @sprintf("%.2f%%", (treynor_B) * 100), "  Jensen alpha=", @sprintf("%.2f%%", (jensen_B) * 100), "  Sharpe=", @sprintf("%.2f", sharpe_B)))
```

**3.** Compare the two managers on all three measures. Do all three measures agree on who performed better?

```julia
println("Manager          Treynor     Jensen     Sharpe")
println("Manager A      ", @sprintf("%10.2f%% %10.2f%% %10.2f", treynor_A * 100, jensen_A * 100, sharpe_A))
println("Manager B      ", @sprintf("%10.2f%% %10.2f%% %10.2f", treynor_B * 100, jensen_B * 100, sharpe_B))
println("B wins on Treynor & Jensen; A wins on Sharpe -- a real ranking reversal.")
```

## Step 4 — Julia Pass *(7 minutes)*

- All three performance measures (Sharpe, Treynor, Jensen's alpha), built from return/σ/beta cells for both managers
- A summary block ranking both managers under each measure side by side
- A short flag noting anywhere the rankings disagree

```julia
function evaluate_manager(rp, sigma, beta, r_f, r_m)
    treynor = (rp - r_f) / beta
    jensen = rp - (r_f + beta * (r_m - r_f))
    sharpe = (rp - r_f) / sigma
    return Dict(:treynor => treynor, :jensen => jensen, :sharpe => sharpe)
end

A = evaluate_manager(rp_A, sigma_A, beta_A, r_f, r_m)
B = evaluate_manager(rp_B, sigma_B, beta_B, r_f, r_m)
for measure in keys(A)
    winner = A[measure] > B[measure] ? "A" : "B"
    println("$(measure): A=$(round(A[measure]; digits=3))  B=$(round(B[measure]; digits=3))  -> $(winner) wins")
end
```
