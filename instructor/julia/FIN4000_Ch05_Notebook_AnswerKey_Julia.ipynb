## Step 2 — Worked Example *(10 minutes, instructor-led)*

Five years of returns for Stock A and T-bills.

```julia
returns_A = [0.08, -0.03, 0.22, 0.15, -0.10]
returns_T = [0.02,  0.02, 0.03, 0.03,  0.04]
```

```julia
avg_A = mean(returns_A)
std_A = returns_A |> std
println(string("Avg=", @sprintf("%.2f%%", (avg_A) * 100), "  Std=", @sprintf("%.2f%%", (std_A) * 100)))
```

## Step 4 — Julia Pass *(7 minutes)*

- AVERAGE() and STDEV() for Stock A and T-bills
- A risk-premium formula that references those cells
- A Sharpe-ratio formula built from the cells above — not retyped numbers

```julia
function summarize_returns(returns, riskfree)
    avg = mean(returns)
    sample_std = std(returns)
    rp = avg - mean(riskfree)
    return Dict(:avg => avg, :std => sample_std, :risk_premium => rp, :sharpe => rp / sample_std)
end

summarize_returns(returns_A, returns_T)
```
