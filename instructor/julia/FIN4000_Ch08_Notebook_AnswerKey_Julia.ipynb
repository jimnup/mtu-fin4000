## Step 2 — Worked Example *(10 minutes, instructor-led)*

A stock has beta 1.3. The market's standard deviation is 18%. The stock's total standard deviation is 25%. Decompose its variance into systematic and firm-specific components.

```julia
beta1, sigma_m, sigma_total1 = 1.3, 0.18, 0.25
```

```julia
sys_var1 = beta1^2 * sigma_m^2
firm_var1 = sigma_total1^2 - sys_var1
R2_1 = sys_var1 / sigma_total1^2
println(string("Systematic sd=", @sprintf("%.2f%%", (sqrt.(sys_var1)) * 100), "  Firm-specific sd=", @sprintf("%.2f%%", (sqrt.(firm_var1)) * 100), "  R2=", @sprintf("%.2f%%", (R2_1) * 100)))
```

*Systematic variance = 1.3²×0.18² = 0.0548 (σ=23.40%). Total variance = 0.25²=0.0625. Firm-specific variance = 0.0625−0.0548=0.0077 (σ=8.80%). Most of this stock's risk is systematic — diversification won't remove much of it. R² (proportion systematic) = 87.6%.*

## Step 4 — Julia Pass *(7 minutes)*

- The systematic/firm-specific variance decomposition, built from beta/σₘ/σ-total cells so any stock can be tested
- R² computed automatically alongside the decomposition
- The same decomposition run side by side for both stocks in the pair-work question

```julia
function decompose_risk(beta, sigma_m, sigma_total)
    sys_var = beta^2 * sigma_m^2
    firm_var = sigma_total^2 - sys_var
    return Dict(:systematic_sd => sqrt(sys_var), :firm_specific_sd => sqrt(firm_var), :R2 => sys_var / sigma_total^2)
end

println(decompose_risk(1.3, 0.18, 0.25))
println(decompose_risk(0.7, 0.18, 0.25))
```
