## Step 2 — Worked Example *(10 minutes, instructor-led)*

An industry's sales historically grow 1.8× as fast as GDP growth (its “GDP elasticity” is 1.8). Forecast the industry's sales growth in an expansion (GDP +3%) and a recession (GDP −2%).

```julia
elasticity_cyclical = 1.8
gdp_expansion, gdp_recession = 0.03, -0.02
```

```julia
sales_exp = elasticity_cyclical * gdp_expansion
sales_rec = elasticity_cyclical * gdp_recession
println(string("Expansion=", @sprintf("%.2f%%", (sales_exp) * 100), "  Recession=", @sprintf("%.2f%%", (sales_rec) * 100)))
```

*Expansion: 1.8×3%=5.4% sales growth. Recession: 1.8×(−2%)=−3.6% (an actual sales decline). An elasticity above 1.0 means this industry amplifies the business cycle — it's cyclical, not defensive.*

## Step 3 — Pair Work *(12 minutes, with a partner)*

**1.** A defensive industry (e.g., utilities) has a GDP elasticity of 0.4. Forecast its sales growth in both the expansion and recession scenarios. Compare the swing to the cyclical industry's.

```julia
elasticity_defensive = 0.4
sales_exp_def = elasticity_defensive * gdp_expansion
sales_rec_def = elasticity_defensive * gdp_recession
println("Defensive: Expansion=$(round(sales_exp_def * 100; digits=2))%  Recession=$(round(sales_rec_def * 100; digits=2))%")
```

## Step 4 — Julia Pass *(7 minutes)*

- The elasticity-based sales forecast, built from elasticity/GDP-growth cells so any industry or scenario can be tested
- Both industries (cyclical and defensive) forecast side by side across both scenarios
- A quick swing calculation: expansion forecast minus recession forecast, to quantify each industry's cyclicality in one number

```julia
swing_cyclical = sales_exp - sales_rec
swing_defensive = sales_exp_def - sales_rec_def
println("Cyclical swing=$(round(swing_cyclical * 100; digits=2))%   Defensive swing=$(round(swing_defensive * 100; digits=2))%")
```
