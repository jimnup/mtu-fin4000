## Step 2 — Worked Example *(10 minutes, instructor-led)*

Three stocks make up a small index. Stock Y then does a 4-for-1 split.

```julia
# Given: a 3-stock index, before and after a 4-for-1 split in Stock Y
prices_day0  = [20, 80, 40]   # X, Y, Z
shares_day0  = [10,  2,  5]   # millions
prices_after = [20, 20, 40]   # Y split 4-for-1: $80 -> $20
shares_after = [10,  8,  5]   # Y's shares: 2M -> 8M
```

**Price-weighted index** — average of the raw prices (divisor = 3, unadjusted):

```julia
pw_day0  = mean(prices_day0)
pw_after = mean(prices_after)
println(string("PW day0=", @sprintf("%.2f", pw_day0), "   PW after (distorted)=", @sprintf("%.2f", pw_after)))
```

**Value-weighted index** — based on total market cap, which a split doesn't change:

```julia
mktcap_day0  = sum(prices_day0 .* shares_day0)
mktcap_after = sum(prices_after .* shares_after)
vw_after = mktcap_after / mktcap_day0 * 100
println(string("Market cap day0=", string(mktcap_day0), "  after=", string(mktcap_after), "  VW index after=", @sprintf("%.2f", vw_after)))
```

**2.** Compute the value-weighted index level right after the split. What changed, and why?

```julia
face, price, days = 10000, 9800, 91
bank_discount_yield = (face - price) / face * (360 / days)
bond_equiv_yield    = (face - price) / price * (365 / days)
println(string("Bank discount yield=", @sprintf("%.4f%%", (bank_discount_yield) * 100), "  BEY=", @sprintf("%.4f%%", (bond_equiv_yield) * 100)))
```

## Step 4 — Julia Pass *(7 minutes)*

Wrap the two index calculations into reusable functions:

- The price-weighted index and its divisor adjustment after the split
- The value-weighted index (market cap Day 0 vs. after split)
- Both T-bill yield formulas from one Face/Price/Days input block

```julia
function price_weighted(prices, divisor=nothing)
    return sum(prices) / (divisor === nothing ? length(prices) : divisor)
end

function value_weighted(prices, shares, base_prices, base_shares)
    return sum(prices .* shares) / sum(base_prices .* base_shares) * 100
end

println(price_weighted(prices_after, new_divisor))
println(value_weighted(prices_after, shares_after, prices_day0, shares_day0))
```
