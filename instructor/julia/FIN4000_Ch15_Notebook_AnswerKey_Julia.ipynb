## Step 2 — Worked Example *(10 minutes, instructor-led)*

Spot rates (zero-coupon yields) for Years 1-3.

```julia
y1, y2, y3 = 0.04, 0.05, 0.055
```

```julia
f2 = (1 + y2)^2 / (1 + y1) - 1
println(string("f2 = ", @sprintf("%.2f%%", (f2) * 100)))
```

*f₂ = (1.05)² / 1.04 − 1 = 6.01%. Since the curve slopes upward, the implied forward rate is higher than either spot rate — the market is pricing in higher short-term rates ahead.*

## Step 3 — Pair Work *(12 minutes, with a partner)*

```julia
f3 = (1 + y3)^3 / (1 + y2)^2 - 1
println(string("f3 = ", @sprintf("%.2f%%", (f3) * 100)))
```

**2.** Price a 3-year, 6% annual-coupon, $1,000 face-value bond by discounting each cash flow at its own maturity-matched spot rate.

```julia
cash_flows = [60, 60, 1060]
spot_rates = [y1, y2, y3]
periods = [1, 2, 3]
price_spot = sum(cash_flows ./ (1 .+ spot_rates) .^ periods)
println(string("Price (spot-rate discounting) = \$", @sprintf("%.2f", price_spot)))
```

**3.** Now price the same bond using a single flat rate equal to the 3-year spot rate (5.5%) for all three cash flows. How does this price compare to the one from spot-rate discounting?

```julia
function price_with_spot_rates(cash_flows, rates, periods)
    return sum(cash_flows ./ (1 .+ rates) .^ periods)
end

function price_with_flat_rate(cash_flows, flat_rate, periods)
    return sum(cash_flows ./ (1 + flat_rate) .^ periods)
end

println(price_with_spot_rates(cash_flows, spot_rates, periods))
println(price_with_flat_rate(cash_flows, y3, periods))
```
