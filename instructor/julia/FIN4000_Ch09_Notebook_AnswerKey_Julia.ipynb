## Step 2 — Worked Example *(10 minutes, instructor-led)*

The risk-free rate is 4%, the market risk premium is 6%. A stock has beta 1.4. Find its CAPM-required return, and compare it to the stock's own forecast return of 12%.

```julia
r_f, mrp = 0.04, 0.06
beta1, forecast1 = 1.4, 0.12
```

```julia
required1 = r_f + beta1 * mrp
alpha1 = forecast1 - required1
println(string("Required=", @sprintf("%.2f%%", (required1) * 100), "  Alpha=", @sprintf("%.2f%%", (alpha1) * 100)))
```

*Required return = 4%+1.4×6%=12.4%. Alpha = 12%−12.4% = −0.4%. This stock is forecast to earn slightly LESS than the CAPM says it should for its risk — it appears slightly overpriced.*

## Step 3 — Pair Work *(12 minutes, with a partner)*

**1.** A second stock has beta 0.8 and a forecast return of 10%. Find its required return and alpha. Is it overpriced or underpriced?

```julia
beta2, forecast2 = 0.8, 0.10
required2 = r_f + beta2 * mrp
alpha2 = forecast2 - required2
verdict = alpha2 > 0 ? "underpriced" : "overpriced"
println("Required=$(round(required2 * 100; digits=2))%  Alpha=$(round(alpha2 * 100; digits=2))% -> $verdict")
```
